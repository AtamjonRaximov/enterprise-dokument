## Distribution of numerical features




```python
# continuos data taqsimoti histogram orqali ko'riladi
sns.histplot(df['traffic_volume'], kde=True)
plt.grid()
plt.show()
```

```python
# targetni normal distributioni ko'ramiz
x = df['traffic_volume'].sort_values()
mu  = x.mean()
sigma = x.std()
data = norm.pdf(x,mu,sigma)
sns.lineplot(x=x,y=data)
plt.grid()
plt.show()
```

### target deyarli normal taqsimlangan va o'ngga og'gan

```python
## iqr metodini ishlatamiz
q1 = df['traffic_volume'].quantile(0.25)
q3 = df['traffic_volume'].quantile(0.75)
iqr = q3-q1
lower_bound = q1 - 1.5*iqr
upper_bound = q3 + 1.5*iqr
#filterlaymiz
mask1 = df['traffic_volume'] > lower_bound
mask2 = df['traffic_volume'] < upper_bound
mask = mask1 & mask2
df_new = df[mask].copy()
```

```python
df_new.shape
```

```python
# targetni normal distributioni ko'ramiz
x = df_new['traffic_volume'].sort_values()
mu  = x.mean()
sigma = x.std()
data = norm.pdf(x,mu,sigma)
plt.plot(x,data)
plt.grid()
plt.show()
```

```python
#z score metodini ishlatamiz
mu = df_new['traffic_volume'].mean()
sigma = df_new['traffic_volume'].std()
z  = (df_new['traffic_volume']-mu)/sigma
# chegaralarni qo'yamiz va filtirlaymiz
mask1 = z> -3
mask2 = z < 3
mask = mask1 & mask2
df_new = df_new[mask].copy()
df_new.shape
```

```python
## loglaymiz
df_new['traffic_volume_log'] = np.log(df_new['traffic_volume']+1)


# targetni normal distributioni ko'ramiz
x = df_new['traffic_volume_log'].sort_values()
mu  = x.mean()
sigma = x.std()
data = norm.pdf(x,mu,sigma)

plt.plot(x,data)
plt.grid()
plt.show()
```
